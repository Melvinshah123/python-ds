```python
# Arithmetic operations
a = 10
b = 15
add_result = a + b
sub_result = a - b
mul_result = a * b
div_result = a / b
floor_div_result = a // b
modulus_result = a % b
exponential_result = a ** b

print(add_result)
print(sub_result)
print(mul_result)
print(div_result)
print(floor_div_result)
print(modulus_result)
print(exponential_result)
```

```python
# Division vs Floor Divison
print(21 / 5)
print(21 // 5)
```

#### Comparison Operators

```python
# Comparison operators
# == equal to 

a = 10
b = 10

print(a==b)
```

```python
str1 = 'Krish'
str2 = 'krish'

print(str1 == str2)
```

```python
# Not Equal To 
str1!=str2
```

```python
# Greater than and less than
num1 = 45
num2 = 55
num3 = 10
num4 = 40

print(num2 > num1)
print(num3 < num4)
```

```python
# greater than or equal to , less than or equal to
number1 = 45
number2 = 45
number3 = 44

print(number1 >= number2)
print(number1 <= number3)
```

#### Logical Operators

```python
# And , Or , Not
x = True
y = False

print(x and y)
print(x or y)
print(not x)
```
